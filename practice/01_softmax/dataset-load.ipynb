```python
#引入
from torchvision import transforms
from torchvision.datasets import FashionMNIST
from torch.utils.data import DataLoader
import d2l
```

```python
# 转换为tensor
trans = transforms.ToTensor()

# 数据集下载
train_set = FashionMNIST(
    root='../data',
    train=True,
    transform=trans,
    download=True
)

test_set = FashionMNIST(
    root='../data',
    train=False,
    transform=trans,
    download=True
)
```

```python
# 数据加载
batch = 256
workers = 4

train_iter = DataLoader(
    train_set,
    batch_size=batch,
    shuffle=True,
    num_workers=workers
)

test_iter = DataLoader(
    test_set,
    batch_size=batch,
    shuffle=False,
    num_workers=workers
)
```
